```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#Importing Datasets
df = pd.read_csv('../data/athlete_events.csv')
noc = pd.read_csv('../data/noc_regions.csv')
```

```python
#Filtering Summer Olympics
df = df[df['Season'] == 'Summer']
```

```python
df = df.merge(noc, on='NOC', how='left')
```

```python
#Dropping Duplicate Rows
df.drop_duplicates(inplace=True)
```

```python
#All unique regions
df['region'].unique()
```

```python
#Encoding medal distribution
df = pd.concat([df, pd.get_dummies(df['Medal'], dtype='int')], axis = 1)
```

```python
#Grouping countries by medals
region_medal_tally = (
    df.groupby('NOC')[['Gold', 'Silver', 'Bronze']]
    .sum()
    .sort_values('Gold', ascending = False)
    .reset_index()
)
```

```python
#Years List
years = df['Year'].unique().tolist()
years.sort()
years.insert(0, 'Overall')
print(years)
```

```python
#Country List
country = np.unique(df['region'].dropna().values).tolist()
country.sort()
country.insert(0, 'Overall')
print(country)
```

```python
#Dropping duplicated medal records
medal_df = df.drop_duplicates(subset=['Team', 'NOC', 'Games', 'Year', 'City', 'Sport', 'Event', 'Medal'])
```

```python
#Flitered Medal Tally
def medal_tally(year, country):
    flag = 0

    if (year == 'Overall' and country == 'Overall'):
        flag = 1
    elif (year == 'Overall' and country != 'Overall'):
        flag = 2
        temp_df = medal_df[medal_df['region'] == country]
    elif (year != 'Overall' and country == 'Overall'):
        flag = 3
        temp_df = medal_df[medal_df['Year'] == year]
    elif (year != 'Overall' and country != 'Overall'):
        flag = 4
        temp_df = medal_df[(medal_df['region'] == country) & (medal_df['Year'] == year)]

    if (flag == 1):
        x = medal_df.groupby('Year')[['Gold', 'Silver', 'Bronze']].sum().sort_values('Year', ascending = False).reset_index()
    elif (flag == 2):
        x = temp_df.groupby('Year')[['Gold', 'Silver', 'Bronze']].sum().sort_values('Year', ascending = False).reset_index()
    elif (flag == 3):
        x = temp_df.groupby('region')[['Gold', 'Silver', 'Bronze']].sum().sort_values('Gold', ascending = False).reset_index()
    elif (flag == 4):
        x = temp_df.groupby('Year')[['Gold', 'Silver', 'Bronze']].sum().sort_values('Year', ascending = False).reset_index()

    return x
```
